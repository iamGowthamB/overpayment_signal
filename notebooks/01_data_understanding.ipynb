## 12. Verify Relationship between Cases and Payments

We check how the datasets align on the primary/foreign key `case_id`.

```python
cases_ids = set(df_cases['case_id'])
payments_ids = set(df_payments['case_id'])
print(f'Unique cases in cases.csv: {len(cases_ids)}')
print(f'Unique cases in payments.csv: {len(payments_ids)}')
print(f'Intersection of cases: {len(cases_ids.intersection(payments_ids))}')
```

## 13. Payment Count Distribution per Case

## Initial Observations

The following observations are directly supported by the data profiling and checking:

1. **Case Status Distribution**: The dataset contains 4,200 benefit cases, where **3,728 (88.76%)** are `Active`, **251 (5.98%)** are `Closed`, and **221 (5.26%)** are `Suspended`.
2. **Payment Temporal Scope**: Payments are tracked for a 6-month window from **2025-07 to 2025-12** (inclusive).
3. **Perfect Case-Payment Alignment**: There is a 1-to-1 linkage on `case_id` between `cases.csv` and `payments.csv`. No cases lack payments, and no payments refer to non-existent cases.
4. **Payments for Closed Cases (Potential Overpayment Signal)**:
   * Cases marked as `Closed` have an average of ~4 payments. However, examining payment dates against `closure_month` reveals that some cases continued to receive payments *after* their closure month.
   * For example, out of cases closed in `2025-08`, some cases continued to receive payments in `2025-09` through `2025-12`. This represents a strong, objective data signal for potential improper overpayments.
5. **Duplicate Case-Month Transactions**: There are 62 instances of a case receiving multiple payment records in a single month (often using different payment methods such as one via Transfer and one via Card for the same amount). This could represent system duplication errors or split disbursements.
6. **Discrepancy Between Awards and Payments**: The payment amount (`amount` in payments) and the declared monthly benefit amount (`monthly_award` in cases) differ in the vast majority of transactions. Only a small fraction (343 out of 24,756) match within $0.01. Further analysis is required to understand if these differences are due to adjustments or other variables.
7. **Fairness Demographics**: District, Age Band, Language Preference, and Housing Tenure have complete data (0% missing) for all 4,200 cases, which will allow for robust demographic parity checks during fairness evaluation.
